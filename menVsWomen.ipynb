```python
import pandas as pd
import plotly.graph_objects as go
import plotly.io as pio

pio.renderers.default = "browser"

# --------------------------------
# Grade scale
# --------------------------------
grade_to_num = {
    "8a": 0,
    "8a+": 1,
    "8b": 2,
    "8b+": 3,
    "8c": 4,
    "8c+": 5,
    "9a": 6,
    "9a/9a+": 6.5,
    "9a+": 7,
    "9b": 8,
    "9b+": 9,
    "9c": 10,
}

tickvals = [0, 1, 2, 3, 4, 5, 6, 6.5, 7, 8, 9, 10]
ticktext = ["8a", "8a+", "8b", "8b+", "8c", "8c+", "9a", "9a/9a+", "9a+", "9b", "9b+", "9c"]

# --------------------------------
# Milestone data
# --------------------------------
men_milestones = [
    (1983, "8a+"),
    (1984, "8b"),
    (1985, "8b+"),
    (1987, "8c"),
    (1990, "8c+"),
    (1991, "9a"),
    (2001, "9a+"),
    (2008, "9b"),
    (2012, "9b+"),
    (2017, "9c"),
]

women_milestones = [
    (1986, "8a"),
    (1988, "8a+"),
    (1988, "8b"),
    (1990, "8b+"),
    (1998, "8c"),
    (2000, "8c+"),
    (2002, "9a"),
    (2005, "9a/9a+"),
    (2017, "9a+"),
    (2017, "9b"),
    (2025, "9b+"),
]

men_df = pd.DataFrame(men_milestones, columns=["year", "grade"])
women_df = pd.DataFrame(women_milestones, columns=["year", "grade"])

men_df["grade_num"] = men_df["grade"].map(grade_to_num)
women_df["grade_num"] = women_df["grade"].map(grade_to_num)

# collapse duplicate milestone years to highest grade reached that year
men_year_max = men_df.groupby("year", as_index=False)["grade_num"].max()
women_year_max = women_df.groupby("year", as_index=False)["grade_num"].max()

start_year = min(men_year_max["year"].min(), women_year_max["year"].min())
end_year = 2026

all_years = pd.DataFrame({"year": range(start_year, end_year + 1)})

# merge and forward-fill so the line is continuous across every year
men_yearly = all_years.merge(men_year_max, on="year", how="left")
women_yearly = all_years.merge(women_year_max, on="year", how="left")

men_yearly["frontier"] = men_yearly["grade_num"].ffill()
women_yearly["frontier"] = women_yearly["grade_num"].ffill()

# --------------------------------
# Plot
# --------------------------------
fig = go.Figure()

fig.add_trace(go.Scatter(
    x=men_yearly["year"],
    y=men_yearly["frontier"],
    mode="lines",
    name="Men",
    line=dict(width=6),
    line_shape="hv",
))

fig.add_trace(go.Scatter(
    x=women_yearly["year"],
    y=women_yearly["frontier"],
    mode="lines",
    name="Women",
    line=dict(width=6),
    line_shape="hv",
))

# optional milestone markers
fig.add_trace(go.Scatter(
    x=men_year_max["year"],
    y=men_year_max["grade_num"],
    mode="markers+text",
    name="Men milestones",
    text=[k for k, v in grade_to_num.items() if False],  # placeholder, replaced below
    showlegend=False,
))

fig.data[-1].text = [
    next(g for g, n in grade_to_num.items() if n == val)
    for val in men_year_max["grade_num"]
]
fig.data[-1].textposition = "top center"
fig.data[-1].marker = dict(size=8)

fig.add_trace(go.Scatter(
    x=women_year_max["year"],
    y=women_year_max["grade_num"],
    mode="markers+text",
    name="Women milestones",
    text=[
        next(g for g, n in grade_to_num.items() if n == val)
        for val in women_year_max["grade_num"]
    ],
    textposition="bottom center",
    marker=dict(size=8),
    showlegend=False,
))

fig.update_layout(
    title="Highest Sport Grade Redpointed by Men vs Women Over Time",
    template="plotly_dark",
    xaxis=dict(
        title="Year",
        range=[start_year - 1, end_year + 1],
    ),
    yaxis=dict(
        title="Highest Grade Achieved",
        tickmode="array",
        tickvals=tickvals,
        ticktext=ticktext,
        range=[-0.3, 10.5],
    ),
    legend=dict(
        title="Category",
        x=1.02,
        y=1,
        xanchor="left",
        yanchor="top",
    ),
    margin=dict(r=180),
    width=1200,
    height=700,
)

fig.show(renderer="browser")
```

```python
import pandas as pd
import plotly.graph_objects as go
import plotly.io as pio

pio.renderers.default = "browser"

# --------------------------------
# Grade scale
# --------------------------------
grade_to_num = {
    "7b+": 0,
    "7c": 1,
    "7c+": 2,
    "8a": 3,
    "8a/+": 3.5,
    "8a+": 4,
    "8b": 5,
    "8b+": 6,
    "8c": 7,
    "8c+": 8,
    "9a": 9,
    "9a+": 10,
}

# --------------------------------
# Earliest flash/onsight for each grade
# using whichever came first
# --------------------------------
men_milestones = [
    (1982, "7b+"),
    (1982, "7c"),
    (1984, "7c+"),
    (1987, "8a"),
    # 8a+ unknown -> omitted
    (1993, "8b"),
    (1995, "8b+"),
    (2004, "8c"),
    (2007, "8c+"),
    (2012, "9a"),
    (2018, "9a+"),
]

women_milestones = [
    (1992, "8a"),
    (1993, "8a/+"),
    (2000, "8a+"),
    (2004, "8b"),
    (2006, "8b+"),
    (2015, "8c"),
    (2025, "8c+"),
]

men_df = pd.DataFrame(men_milestones, columns=["year", "grade"])
women_df = pd.DataFrame(women_milestones, columns=["year", "grade"])

men_df["grade_num"] = men_df["grade"].map(grade_to_num)
women_df["grade_num"] = women_df["grade"].map(grade_to_num)

# --------------------------------
# Highest grade reached in each year
# --------------------------------
men_year_max = men_df.groupby("year", as_index=False)["grade_num"].max()
women_year_max = women_df.groupby("year", as_index=False)["grade_num"].max()

start_year = min(men_year_max["year"].min(), women_year_max["year"].min())
end_year = 2026

all_years = pd.DataFrame({"year": range(start_year, end_year + 1)})

men_yearly = all_years.merge(men_year_max, on="year", how="left").sort_values("year")
women_yearly = all_years.merge(women_year_max, on="year", how="left").sort_values("year")

# cumulative frontier = highest grade achieved up to that point
men_yearly["frontier"] = men_yearly["grade_num"].cummax().ffill()
women_yearly["frontier"] = women_yearly["grade_num"].cummax().ffill()

# --------------------------------
# Gap series
# --------------------------------
gap_df = all_years.copy()
gap_df["men_frontier"] = men_yearly["frontier"]
gap_df["women_frontier"] = women_yearly["frontier"]
gap_df["gap"] = gap_df["men_frontier"] - gap_df["women_frontier"]

# only plot where both frontiers exist
gap_plot = gap_df.dropna(subset=["gap"]).copy()

# optional labels at milestone years
milestone_years = sorted(set(men_year_max["year"]).union(set(women_year_max["year"])))
label_df = gap_plot[gap_plot["year"].isin(milestone_years)].copy()

# --------------------------------
# Plot
# --------------------------------
fig = go.Figure()

fig.add_trace(go.Scatter(
    x=gap_plot["year"],
    y=gap_plot["gap"],
    mode="lines",
    name="Gap",
    line=dict(width=6),
    line_shape="hv",
    fill="tozeroy",
    hovertemplate="Year %{x}<br>Gap: %{y:g} grades<extra></extra>",
))

fig.add_trace(go.Scatter(
    x=label_df["year"],
    y=label_df["gap"],
    mode="markers+text",
    text=[f"{v:g}" for v in label_df["gap"]],
    textposition="top center",
    marker=dict(size=8),
    showlegend=False,
    hovertemplate="Year %{x}<br>Gap: %{y:g} grades<extra></extra>",
))

fig.update_layout(
    title="Gap Between Highest Sport Flash/Onsight Grade Achieved by Men and Women Over Time",
    template="plotly_dark",
    xaxis=dict(
        title="Year",
        range=[gap_plot["year"].min() - 1, end_year + 1],
    ),
    yaxis=dict(
        title="Gap (grades)",
        range=[0, gap_plot["gap"].max() + 1],
        dtick=0.5,
    ),
    width=1250,
    height=650,
)

fig.show(renderer="browser")
```
