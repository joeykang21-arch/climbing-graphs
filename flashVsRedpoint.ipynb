```python
import pandas as pd
import plotly.graph_objects as go
import plotly.io as pio

pio.renderers.default = "browser"

# --------------------------------
# Grade scale
# --------------------------------
grade_to_num = {
    "7b+": 0,
    "7c": 1,
    "7c+": 2,
    "8a": 3,
    "8a+": 4,
    "8b": 5,
    "8b+": 6,
    "8c": 7,
    "8c+": 8,
    "9a": 9,
    "9a+": 10,
    "9b": 11,
    "9b+": 12,
    "9c": 13,
}

num_to_grade = {v: k for k, v in grade_to_num.items()}

# --------------------------------
# Men redpoint milestones
# --------------------------------
redpoint_milestones = [
    (1983, "8a+"),
    (1984, "8b"),
    (1985, "8b+"),
    (1987, "8c"),
    (1990, "8c+"),
    (1991, "9a"),
    (2001, "9a+"),
    (2008, "9b"),
    (2012, "9b+"),
    (2017, "9c"),   # treating Silence as 9c
]

# --------------------------------
# Men flash/onsight milestones
# earliest of either style for each grade
# 8a+ unknown -> omitted
# --------------------------------
flash_onsight_milestones = [
    (1982, "7b+"),
    (1982, "7c"),
    (1984, "7c+"),
    (1987, "8a"),
    (1993, "8b"),
    (1995, "8b+"),
    (2004, "8c"),
    (2007, "8c+"),
    (2012, "9a"),
    (2018, "9a+"),
]

# --------------------------------
# Build DataFrames
# --------------------------------
red_df = pd.DataFrame(redpoint_milestones, columns=["year", "grade"])
fo_df = pd.DataFrame(flash_onsight_milestones, columns=["year", "grade"])

red_df["grade_num"] = red_df["grade"].map(grade_to_num)
fo_df["grade_num"] = fo_df["grade"].map(grade_to_num)

# --------------------------------
# Highest grade reached in each milestone year
# --------------------------------
red_year_max = red_df.groupby("year", as_index=False)["grade_num"].max()
fo_year_max = fo_df.groupby("year", as_index=False)["grade_num"].max()

start_year = min(red_year_max["year"].min(), fo_year_max["year"].min())
end_year = 2026

all_years = pd.DataFrame({"year": range(start_year, end_year + 1)})

red_yearly = all_years.merge(red_year_max, on="year", how="left").sort_values("year")
fo_yearly = all_years.merge(fo_year_max, on="year", how="left").sort_values("year")

# cumulative frontier
red_yearly["frontier_redpoint"] = red_yearly["grade_num"].cummax().ffill()
fo_yearly["frontier_flash_onsight"] = fo_yearly["grade_num"].cummax().ffill()

# --------------------------------
# Merge frontiers and compute gap
# --------------------------------
frontier_df = all_years.copy()
frontier_df["redpoint"] = red_yearly["frontier_redpoint"]
frontier_df["flash_onsight"] = fo_yearly["frontier_flash_onsight"]

# only keep years where both exist
frontier_df = frontier_df.dropna(subset=["redpoint", "flash_onsight"]).copy()

frontier_df["gap"] = frontier_df["redpoint"] - frontier_df["flash_onsight"]

# Optional: label the exact grades at each year
frontier_df["redpoint_grade"] = frontier_df["redpoint"].map(num_to_grade)
frontier_df["flash_grade"] = frontier_df["flash_onsight"].map(num_to_grade)

# --------------------------------
# Plot gap over time
# --------------------------------
gap_fig = go.Figure()

gap_fig.add_trace(go.Scatter(
    x=frontier_df["year"],
    y=frontier_df["gap"],
    mode="lines",
    line_shape="hv",
    line=dict(width=5),
    fill="tozeroy",
    hovertemplate=(
        "Year: %{x}<br>"
        "Gap: %{y} grade steps<br>"
        "Redpoint: %{customdata[0]}<br>"
        "Flash/Onsight: %{customdata[1]}"
        "<extra></extra>"
    ),
    customdata=frontier_df[["redpoint_grade", "flash_grade"]].values,
    name="Gap",
))

gap_fig.update_layout(
    title="Gap Between Hardest Redpoint and Hardest Flash/Onsight Over Time",
    template="plotly_dark",
    xaxis=dict(
        title="Year",
        range=[frontier_df["year"].min() - 1, end_year + 1],
    ),
    yaxis=dict(
        title="Gap (grade steps)",
        dtick=1,
    ),
    width=1100,
    height=500,
)

gap_fig.show(renderer="browser")
```

```python
import numpy as np
import pandas as pd
import plotly.graph_objects as go
import plotly.io as pio

pio.renderers.default = "browser"

# ============================
# 🎛️ THUMBNAIL TUNING
# ============================
RANDOM_SEED = 7
HIGHLIGHT_NAME = "Adam Ondra"

# Canvas size
FIG_WIDTH = 1728
FIG_HEIGHT = 972

# Adam placement
ADAM_X = 0.18
ADAM_Y = 110

# Named climbers you want visible
NAMED_CLIMBERS = [
    ("Will Bosi", 28),
    ("Jimmy Webb", 21),
    ("Yannick Flohé", 19),
    ("Daniel Woods", 15),
    ("Aidan Roberts", 13),
    ("Nicolai Uznik", 12),
    ("Jakob Schubert", 19),
    ("Tomoa Narasaki", 10),
    ("Noah Wheeler", 7),
    ("Jules Marchaland", 20),
    ("Vadim Timonov", 17),
]

# Manual x placements for readable thumbnail layout
MANUAL_X = {
    "Vadim Timonov": -0.44,
    "Jules Marchaland": -0.22,
    "Jakob Schubert": -0.08,
    "Noah Wheeler": -0.13,
    "Tomoa Narasaki": -0.05,
    "Nicolai Uznik": 0.12,
    "Jimmy Webb": 0.40,
    "Daniel Woods": 0.42,
    "Aidan Roberts": 0.46,
    "Yannick Flohé": 0.49,
    "Will Bosi": 0.54,
}

# Small bottom crowd
NUM_EXTRA_POINTS = 85
BOTTOM_MIN_Y = 1
BOTTOM_MAX_Y = 6
BOTTOM_ROWS = 3
BOTTOM_X_JITTER = 0.008

OUT_HTML = "adam_ondra_thumbnail_fake_data.html"
# ============================

rng = np.random.default_rng(RANDOM_SEED)

# ----------------------------
# Build forged dataset
# ----------------------------
rows = []

# Adam
rows.append({
    "Climber": HIGHLIGHT_NAME,
    "Score": ADAM_Y,
    "x": ADAM_X,
    "label": HIGHLIGHT_NAME,
    "textpos": "middle right",
    "kind": "highlight"
})

# Named climbers
for name, score in NAMED_CLIMBERS:
    rows.append({
        "Climber": name,
        "Score": score,
        "x": MANUAL_X.get(name, 0),
        "label": name,
        "textpos": "middle right" if MANUAL_X.get(name, 0) <= 0 else "middle left",
        "kind": "named"
    })

# Extra unlabeled crowd near the bottom
x_positions = np.linspace(-0.46, 0.56, NUM_EXTRA_POINTS)

for i in range(NUM_EXTRA_POINTS):
    row = i % BOTTOM_ROWS
    base_y = rng.integers(BOTTOM_MIN_Y, BOTTOM_MAX_Y + 1)
    y = base_y + row * 0.15
    x = x_positions[i] + rng.uniform(-BOTTOM_X_JITTER, BOTTOM_X_JITTER)

    rows.append({
        "Climber": f"Extra {i+1}",
        "Score": float(y),
        "x": float(x),
        "label": "",
        "textpos": "middle center",
        "kind": "crowd"
    })

rank = pd.DataFrame(rows)

# Split highlight vs others
others = rank[rank["kind"] != "highlight"].copy()
hilite = rank[rank["kind"] == "highlight"].copy()

# ----------------------------
# Plot
# ----------------------------
fig = go.Figure()

# Everyone else
fig.add_trace(go.Scatter(
    x=others["x"],
    y=others["Score"],
    mode="markers+text",
    text=others["label"],
    textposition=others["textpos"],
    textfont=dict(size=16, color="white"),
    hovertemplate="<b>%{customdata[0]}</b><br>Score: %{y}<extra></extra>",
    customdata=np.stack([others["Climber"]], axis=-1),
    marker=dict(
        size=9,
        opacity=0.9,
        color="#636EFA"
    ),
    showlegend=False,
))

# Adam
fig.add_trace(go.Scatter(
    x=hilite["x"],
    y=hilite["Score"],
    mode="markers+text",
    text=hilite["label"],
    textposition="middle right",
    textfont=dict(size=22, color="white"),
    hovertemplate="<b>%{customdata[0]}</b><br>Score: %{y}<extra></extra>",
    customdata=np.stack([hilite["Climber"]], axis=-1),
    marker=dict(
        size=16,
        opacity=1.0,
        symbol="diamond",
        color="tomato"
    ),
    showlegend=False,
))

# Left vertical axis line
fig.add_shape(
    type="line",
    x0=0.02, x1=0.02,
    y0=0, y1=1,
    xref="paper",
    yref="paper",
    line=dict(width=3, color="white"),
)

fig.update_layout(
    template="plotly_dark",
    width=FIG_WIDTH,
    height=FIG_HEIGHT,
    xaxis=dict(
        title=None,
        showticklabels=False,
        showgrid=False,
        zeroline=False,
        visible=False,
        range=[-0.50, 0.58],
    ),
    yaxis=dict(
        title=None,
        showgrid=False,
        zeroline=False,
        tickfont=dict(size=18, color="white"),
        range=[0, 115],
    ),
    margin=dict(l=80, r=30, t=30, b=40),
    showlegend=False,
    plot_bgcolor="black",
    paper_bgcolor="black",
)

fig.show()
fig.write_html(OUT_HTML)
print("Saved:", OUT_HTML)
```
